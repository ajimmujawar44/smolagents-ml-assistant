# Create your own custom tool

```python
@tool
def calculate_compound_interest(
    principal: float,
    rate: float,
    years: int
) -> float:
    """
    Calculate compound interest.

    Args:
        principal: Initial investment amount.
        rate: Annual interest rate in percentage.
        years: Number of years.

    Returns:
        Final amount after compound interest.
    """
    
    final_amount = principal * (1 + rate / 100) ** years
    
    return round(final_amount, 2)
```

# Test the custom tool directly

```python
result = calculate_compound_interest(
    principal=10000,
    rate=8,
    years=5
)

print(f"Final amount: ₹{result}")
```

# Give the custom tool to the agent

```python
finance_agent = ToolCallingAgent(
    tools=[calculate_compound_interest],
    model=model,
    max_steps=5
)

print("Finance agent created.")
```

# Ask the agent to use the tool

```python
response = finance_agent.run(
    "If I invest ₹50,000 at an annual interest rate of 7% "
    "for 5 years, calculate the final amount."
)

print(response)
```

# Multiple tools

```python
@tool
def add_numbers(a: float, b: float) -> float:
    """
    Add two numbers.

    Args:
        a: First number.
        b: Second number.

    Returns:
        Sum of the numbers.
    """
    return a + b


@tool
def multiply_numbers(a: float, b: float) -> float:
    """
    Multiply two numbers.

    Args:
        a: First number.
        b: Second number.

    Returns:
        Product of the numbers.
    """
    return a * b
```

# Agent with multiple tools

```python
math_agent = ToolCallingAgent(
    tools=[
        add_numbers,
        multiply_numbers,
        calculate_compound_interest
    ],
    model=model,
    max_steps=6
)

print("Multi-tool agent created.")
```

# Test tool selection

```python
response = math_agent.run(
    "Add 250 and 750, then multiply the result by 3."
)

print(response)
```

# Agent instructions

```python
agent = ToolCallingAgent(
    tools=[search_tool],
    model=model,
    max_steps=6,
    instructions="""
    You are a helpful research assistant.

    Rules:
    1. Understand the user's question first.
    2. Use the search tool when current information is required.
    3. Do not invent facts.
    4. Clearly separate known information from uncertainty.
    5. Give concise and structured answers.
    """
)

print("Instruction-based agent created.")
```

# Structured research task

```python
response = agent.run(
    """
    Research agentic AI.

    Explain:
    1. What agentic AI means
    2. How an AI agent works
    3. Common tools used by agents
    4. Real-world applications
    5. Important limitations
    """
)

print(response)
```

# Error handling

```python
def run_agent_safely(agent, prompt):
    try:
        result = agent.run(prompt)
        return result

    except Exception as e:
        print("Agent execution failed.")
        print(f"Error: {type(e).__name__}")
        print(f"Details: {e}")
        return None
```

# Test safe execution

```python
result = run_agent_safely(
    tool_agent,
    "Explain the difference between AI agents and traditional chatbots."
)

if result:
    print(result)
```

# Build a practical research agent

```python
research_agent = ToolCallingAgent(
    tools=[search_tool],
    model=model,
    max_steps=8,
    instructions="""
    You are an AI research assistant.

    Workflow:
    1. Understand the research question.
    2. Search for relevant information when necessary.
    3. Compare information from the available results.
    4. Avoid unsupported claims.
    5. Produce a concise structured answer.
    6. Mention uncertainty when information is incomplete.
    """
)

print("Research agent ready.")
```

# Final practical test

```python
query = """
Research the current role of Agentic AI in software development.

Explain:
- what AI agents do
- common tools they use
- how developers use them
- important limitations
- skills a developer should learn
"""

result = run_agent_safely(
    research_agent,
    query
)

if result:
    print(result)
```
