## 2. Import libraries

````python
import os
import time
import gradio as gr
import pandas as pd
from dotenv import load_dotenv
from smolagents import CodeAgent, DuckDuckGoSearchTool, LiteLLMModel
````

## 3. Load API key

````python
load_dotenv()
GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY")
if not GOOGLE_API_KEY:
    raise ValueError(
        "GOOGLE_API_KEY was not found. Create a .env file beside this notebook "
        "containing: GOOGLE_API_KEY=your_google_api_key"
    )
print("✅ Google API key loaded.")
````

## 4. Configuration

````python
APP_NAME = "AI ML Code Agent"
AUTHOR_NAME = "Ajim Mujawar"
MODEL_ID = "gemini/gemini-3.8-flash"
MAX_AGENT_STEPS = 12

print(f"👤 Developer: {AUTHOR_NAME}")
print(f"🤖 Model: {MODEL_ID}")
````

## 5. Initialize Gemini model

````python
model = LiteLLMModel(
    model_id=MODEL_ID,
    api_key=GOOGLE_API_KEY,
    temperature=0.2,
    max_tokens=8192,
)
print("✅ Gemini model wrapper initialized.")
````

## 6. Create the agent
The notebook consistently uses `ml_code_agent`; this fixes the original `NameError` from calling an undefined `agent` variable.

````python
ml_code_agent = CodeAgent(
    tools=[DuckDuckGoSearchTool()],
    model=model,
    max_steps=MAX_AGENT_STEPS,
    verbosity_level=1,
    additional_authorized_imports=[
        "pandas", "numpy", "sklearn",
        "sklearn.model_selection", "sklearn.preprocessing", "sklearn.impute",
        "sklearn.metrics", "sklearn.linear_model", "sklearn.ensemble",
        "sklearn.tree", "sklearn.pipeline", "sklearn.compose",
        "sklearn.feature_selection", "json", "os", "pathlib",
    ],
)
print("✅ ML Code Agent created.")
````

## 7. Agent instructions

````python
SYSTEM_INSTRUCTIONS = """
You are an expert Machine Learning Engineering Agent.
Help solve machine-learning and data-science tasks using clear, reproducible Python.

When a CSV dataset is provided:
- Use only supplied column names; never invent columns.
- Identify a likely target only when justified; ask for confirmation if unclear.
- Explain assumptions.
- Handle missing values and categorical features.
- Split data before fitting preprocessing to avoid data leakage.
- Prefer scikit-learn Pipelines and ColumnTransformer where appropriate.
- Choose metrics appropriate to classification or regression.
- Use random_state=42 where applicable.
- Do not fabricate model scores or claim code was executed when it was not.
- Never reveal API keys, credentials, or secrets.

Structure the answer as:
1. Problem understanding
2. Dataset analysis
3. Proposed approach
4. Complete Python code
5. Evaluation plan / metrics
6. Key considerations
7. Next steps
"""
````

## 8. CSV analysis helper

````python
def analyze_csv(csv_path):
    """Return a readable summary of an uploaded CSV file."""
    if not csv_path:
        return "Please upload a CSV file."
    try:
        df = pd.read_csv(csv_path)
        return "\\n".join([
            "### Dataset overview",
            f"- **Rows:** {df.shape[0]:,}",
            f"- **Columns:** {df.shape[1]:,}",
            f"- **Duplicate rows:** {int(df.duplicated().sum()):,}",
            "",
            "### Column names",
            ", ".join(map(str, df.columns)),
            "",
            "### Data types",
            "```text",
            df.dtypes.to_string(),
            "```",
            "",
            "### Missing values",
            "```text",
            df.isnull().sum().to_string(),
            "```",
            "",
            "### Preview (first 5 rows)",
            "```text",
            df.head().to_string(index=False),
            "```",
        ])
    except Exception as exc:
        return f"❌ Could not read this CSV: {exc}"
````

## 9. Run agent with retry

````python
def run_agent_with_retry(task, max_retries=3):
    """Retry temporary service/rate-limit failures; return readable errors."""
    for attempt in range(1, max_retries + 1):
        try:
            return str(ml_code_agent.run(task))
        except Exception as exc:
            message = str(exc)
            transient = any(
                marker in message.lower()
                for marker in ("503", "serviceunavailable", "429", "rate limit", "temporarily unavailable")
            )
            if transient and attempt < max_retries:
                wait_seconds = 2 ** attempt
                print(f"Temporary API issue. Retrying in {wait_seconds}s ({attempt}/{max_retries})...")
                time.sleep(wait_seconds)
                continue
            return (
                "❌ Agent request failed.\\n\\n"
                f"{message}\\n\\n"
                "Check that your API key and model ID are valid, and that your API quota is available."
            )
    return "❌ Agent request failed after retries."
````

## 10. Generate a machine-learning solution

````python
def generate_ml_solution(prompt, csv_path):
    if not csv_path:
        return "❌ Please upload a CSV dataset first."
    if not prompt or not prompt.strip():
        return "❌ Please describe your machine-learning task."

    try:
        df = pd.read_csv(csv_path)
        dataset_info = (
            f"Shape: {df.shape[0]} rows × {df.shape[1]} columns\\n"
            f"Columns: {df.columns.tolist()}\\n\\n"
            f"Data types:\\n{df.dtypes.to_string()}\\n\\n"
            f"Missing values:\\n{df.isnull().sum().to_string()}\\n\\n"
            f"Duplicate rows: {int(df.duplicated().sum())}\\n\\n"
            f"First 5 rows:\\n{df.head().to_string(index=False)}"
        )
        task = f"""
{SYSTEM_INSTRUCTIONS}

DATASET SUMMARY
{dataset_info}

USER TASK
{prompt.strip()}

Create a practical solution based on this dataset summary.
If the target column is unclear, ask for confirmation or state your assumption.
Provide complete, runnable Python code. Do not fabricate model scores or claim
that generated code has been run.
"""
        return run_agent_with_retry(task)
    except Exception as exc:
        return f"❌ Could not prepare the ML task: {exc}"
````

## 11. Example prompts

````python
EXAMPLE_PROMPTS = [
    "Analyze this dataset and explain the columns, data quality, and likely target.",
    "Build a classification pipeline. Ask me to confirm the target if unclear.",
    "Build a regression pipeline and compare suitable algorithms using cross-validation.",
    "Perform exploratory data analysis and suggest useful visualizations.",
    "Handle missing values and categorical features without data leakage.",
    "Explain which evaluation metrics are appropriate for my task.",
]
````

````python
custom_css = """
.gradio-container { max-width: 1250px !important; margin: auto !important; }
#hero { text-align: center; padding: 24px; border-radius: 18px; margin-bottom: 18px; }
#hero h1 { font-size: 34px; margin-bottom: 8px; }
#hero p { font-size: 16px; }
#developer { text-align: center; margin-top: 16px; padding: 12px; opacity: 0.85; }
#generate-btn { min-height: 48px; font-size: 16px; font-weight: 600; }
"""
````

## 13. Build the Gradio app

````python
with gr.Blocks(
    title=f"{APP_NAME} | {AUTHOR_NAME}",
    theme=gr.themes.Soft(),
    css=custom_css,
) as demo:
    gr.HTML("""
    <div id="hero">
        <h1>🤖 AI ML Code Agent</h1>
        <p>Machine-learning assistance powered by <b>SmolAgents + Gemini</b></p>
        <p>Upload a CSV, inspect the dataset, and describe the ML task you want to solve.</p>
    </div>
    """)

    with gr.Row():
        with gr.Column(scale=1):
            gr.Markdown("## 📂 Dataset")
            csv_input = gr.File(
                label="Upload CSV dataset",
                file_types=[".csv"],
                type="filepath",
            )
            analyze_btn = gr.Button("🔍 Analyze dataset")
            dataset_preview = gr.Markdown(
                "Upload a CSV and click **Analyze dataset** to see its summary."
            )

        with gr.Column(scale=2):
            gr.Markdown("## 💬 Machine-learning task")
            prompt_input = gr.Textbox(
                label="Describe your ML task",
                placeholder="Example: Build a classification pipeline and evaluate it using suitable metrics.",
                lines=6,
            )
            gr.Examples(
                examples=EXAMPLE_PROMPTS,
                inputs=prompt_input,
                label="💡 Example tasks",
            )
            generate_btn = gr.Button(
                "🚀 Generate ML solution",
                variant="primary",
                elem_id="generate-btn",
            )

    gr.Markdown("---")
    gr.Markdown("## 🧠 AI-generated solution")
    output = gr.Markdown("Your generated solution will appear here.")

    analyze_btn.click(
        fn=analyze_csv, inputs=[csv_input], outputs=[dataset_preview]
    )
    generate_btn.click(
        fn=generate_ml_solution,
        inputs=[prompt_input, csv_input],
        outputs=[output],
    )

    gr.HTML(f"""
    <div id="developer">
        Built by <b>{AUTHOR_NAME}</b><br>
        Python · Machine Learning · Generative AI
    </div>
    """)
````

## 14. Launch the app

````python
# Run this cell last. share=False gives a local URL only.
demo.launch(inbrowser=True, share=False)
````

## Troubleshooting

- **`NameError: name 'agent' is not defined`:** fixed; the callback now calls `ml_code_agent.run(...)`.
- **API key missing:** put `GOOGLE_API_KEY=your_key` in a `.env` file beside the notebook, restart the kernel, and run all cells from the top.
- **Model not found / unsupported model:** verify which Gemini model IDs are available to your Google API account and update `MODEL_ID`.
- **No public URL:** `share=False` creates a local app URL only. Change it to `share=True` if you need a temporary public Gradio link.
- **Run order:** restart the kernel and select **Run All**. Do not launch the app before defining the functions and `demo`.
