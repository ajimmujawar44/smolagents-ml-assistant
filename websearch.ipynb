```python
from smolagents import CodeAgent, WebSearchAgent, InferenceClientModel

model = InferanceClientModel(moedl_id = "")
agent = CodeAgent(tools=[WebSearchAgent(model=model)], model=model, stream_output=True)

agent.run("Write a python function that takes a string as input and returns the number of vowels in the string. Use web search to find the best implementation.")
```
